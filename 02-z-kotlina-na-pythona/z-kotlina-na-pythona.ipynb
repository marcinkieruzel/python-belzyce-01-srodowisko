## Ściąga: Kotlin → Python

Python to też język z typami, ale nie piszesz ich przy zmiennych. Najważniejsza różnica: **bloki kodu wyznaczają wcięcia (4 spacje), a nie klamry**, a linijka przed blokiem (`if`, `for`, `def`…) kończy się dwukropkiem. Średników nie ma.

| Kotlin | Python | Uwagi |
| --- | --- | --- |
| `val x = 5`, `var x = 5` | `x = 5` | nie ma `val` / `var` ani typu przy zmiennej |
| `x++`, `x += 1` | `x += 1` | w Pythonie nie ma `++` |
| `true`, `false`, `null` | `True`, `False`, `None` | wielka litera |
| `// komentarz` | `# komentarz` | |
| `println("Wynik: $x")` | `print(f"Wynik: {x}")` | f-string: litera `f` przed cudzysłowem |
| `"Suma: ${a + b}"` | `f"Suma: {a + b}"` | w `{ }` może być dowolne wyrażenie |
| `String.format("%6.1f", x)` | `f"{x:6.1f}"` | szerokość 6, jedno miejsce po przecinku |
| `String.format("%4d", n)` | `f"{n:4d}"` | liczba całkowita, szerokość 4 |
| `s.padEnd(12)` | `f"{s:<12}"` | tekst do lewej, szerokość 12 |
| `7 / 2` (dwa `Int`) daje `3` | `7 // 2` daje `3` | w Pythonie `7 / 2` daje `3.5` |
| `a && b`, `a \|\| b`, `!a` | `a and b`, `a or b`, `not a` | |
| `if (x > 0) { … } else { … }` | `if x > 0:` … `else:` … | blok pod spodem z wcięciem |
| `when { x < 0 -> …; x < 10 -> …; else -> … }` | `if x < 0:` … `elif x < 10:` … `else:` … | `elif` to „else if” |
| `for (i in 1..10)` | `for i in range(1, 11):` | **koniec `range` się nie liczy** |
| `for (i in 0 until n)` | `for i in range(n):` | od 0 do n-1 |
| `for (t in -20..40 step 10)` | `for t in range(-20, 41, 10):` | trzeci argument to krok |
| `for (znak in tekst)` | `for znak in tekst:` | |
| `fun pole(a: Int, b: Int): Int { return a * b }` | `def pole(a, b):` + w nowej linii z wcięciem `return a * b` | |
| `fun main() { … }` | — | kod piszesz od razu, bez `main` |
| `listOf(1, 2, 3)`, `mutableListOf<Int>()` | `[1, 2, 3]`, `[]` | lista |
| `lista.add(x)` | `lista.append(x)` | |
| `lista.size`, `tekst.length` | `len(lista)`, `len(tekst)` | |
| `x in lista`, `znak in "!?#"` | `x in lista`, `znak in "!?#"` | tak samo |
| `mutableMapOf<Char, Int>()` | `{}` | słownik |
| `m[k] = v`, `m.getOrDefault(k, 0)` | `m[k] = v`, `m.get(k, 0)` | |
| `for ((k, v) in m)` | `for k, v in m.items():` | |
| `m.toList().sortedByDescending { it.second }` | `sorted(m.items(), key=lambda para: para[1], reverse=True)` | pary posortowane malejąco po drugim elemencie |
| `lista.take(5)` | `lista[:5]` | pierwsze 5 elementów |
| `"#".repeat(n)` | `"#" * n` | |
| `z.isDigit()`, `z.isLetter()`, `z.isUpperCase()` | `z.isdigit()`, `z.isalpha()`, `z.isupper()` | `isalpha()` zna też polskie litery |
| `s.lowercase()`, `s.uppercase()` | `s.lower()`, `s.upper()` | |
| `z in 'a'..'z'` | `"a" <= z <= "z"` | znak to w Pythonie tekst długości 1 |
| `z.code`, `n.toChar()` | `ord(z)`, `chr(n)` | numer znaku i znak o numerze |
| `x.toInt()`, `x.toDouble()`, `x.toString()` | `int(x)`, `float(x)`, `str(x)` | |

Ta sama ściąga w działaniu — uruchom i porównaj z tabelą:

````python
# val imie = "Ola"; var punkty = 0
imie = "Ola"
punkty = 0
punkty += 3

# fun ocena(p: Int): String = when { p >= 3 -> "zdane"; else -> "niezdane" }
def ocena(p):
    if p >= 3:
        return "zdane"
    else:
        return "niezdane"

# println("$imie ma $punkty pkt: ${ocena(punkty)}")
print(f"{imie} ma {punkty} pkt: {ocena(punkty)}")

# for (i in 1..3) println("Runda $i")
for i in range(1, 4):
    print(f"Runda {i}")

# val oceny = listOf(5, 3, 4); println(oceny.size)
oceny = [5, 3, 4]
print(len(oceny), "oceny, średnia:", sum(oceny) / len(oceny))
````

```kotlin
fun opis(c: Int): String {
    return when {
        c < 0 -> "mróz"
        c < 15 -> "chłodno"
        c < 25 -> "przyjemnie"
        else -> "upał"
    }
}

fun main() {

        val k = c + 273.15
        println(String.format("%4d | %6.1f | %6.2f | %s", c, f, k, opis(c)))
    }
}
```

## Zadanie 2. Jak silne jest hasło?

Program w Kotlinie ocenia hasła z listy. Za każdą cechę hasła dostaje punkt: co najmniej 8 znaków, co najmniej 12 znaków, cyfra, wielka litera, znak specjalny. Hasła z listy najpopularniejszych od razu dostają ocenę „fatalne”. Na końcu program wypisuje podsumowanie.

```kotlin
val popularne = listOf("qwerty", "12345678", "haslo123", "zaq12wsx")

fun ocenHaslo(haslo: String): String {
    if (haslo.lowercase() in popularne) {
        return "fatalne"
    }
    var punkty = 0
    if (haslo.length >= 8) punkty++
    if (haslo.length >= 12) punkty++
    var cyfra = false
    var wielka = false
    var specjalny = false
    for (znak in haslo) {
        if (znak.isDigit()) cyfra = true
        if (znak.isUpperCase()) wielka = true
        if (znak in "!@#$%^&*?_-") specjalny = true
    }
    if (cyfra) punkty++
    if (wielka) punkty++
    if (specjalny) punkty++
    return when {
        punkty <= 2 -> "słabe"
        punkty <= 3 -> "średnie"
        else -> "silne"
    }
}

fun main() {
    val hasla = listOf("kot", "Qwerty", "belzyce2026", "Belzyce2026", "Tr0ub4dor&3",
                       "MojeSuperHaslo!2026", "ZAQ12WSX", "dlugiehaslobezcyfr")
    var silne = 0
    for (haslo in hasla) {
        val ocena = ocenHaslo(haslo)
        if (ocena == "silne") silne++
        println("${haslo.padEnd(20)} $ocena")
    }
    println("Silnych haseł: $silne z ${hasla.size}")
}
```

## Zadanie 3. Które litery są najczęstsze?

Program w Kotlinie liczy litery w tekście (bez względu na wielkość, bez spacji i znaków interpunkcyjnych), a potem rysuje histogram pięciu najczęstszych liter ze znaków `#`. Polskie litery liczą się osobno: `ę` to inna litera niż `e`.

```kotlin
fun main() {
    val tekst = "Litwo! Ojczyzno moja! ty jesteś jak zdrowie. " +
        "Ile cię trzeba cenić, ten tylko się dowie, " +
        "kto cię stracił. Dziś piękność twą w całej ozdobie " +
        "widzę i opisuję, bo tęsknię po tobie."

    val licznik = mutableMapOf<Char, Int>()
    var wszystkie = 0
    for (znak in tekst.lowercase()) {
        if (znak.isLetter()) {
            licznik[znak] = licznik.getOrDefault(znak, 0) + 1
            wszystkie++
        }
    }
    println("Liter w tekście: $wszystkie, różnych: ${licznik.size}")

    val top5 = licznik.toList().sortedByDescending { it.second }.take(5)
    for ((litera, ile) in top5) {
        println("$litera ${"#".repeat(ile)} $ile")
    }
}
```

## Zadanie dodatkowe. Szyfr Cezara

Juliusz Cezar szyfrował listy, przesuwając każdą literę o 3 miejsca w alfabecie: `a` → `d`, `b` → `e`, …, `z` → `c`. Program w Kotlinie szyfruje tak tekst złożony z małych liter alfabetu łacińskiego (bez polskich znaków); inne znaki zostawia bez zmian. Ujemne przesunięcie odszyfrowuje.

```kotlin
fun szyfruj(tekst: String, przesuniecie: Int): String {
    var wynik = ""
    for (znak in tekst) {
        if (znak in 'a'..'z') {
            val numer = (znak.code - 'a'.code + przesuniecie + 26) % 26
            wynik += ('a'.code + numer).toChar()
        } else {
            wynik += znak
        }
    }
    return wynik
}

fun main() {
    val tajne = szyfruj("spotkanie o osiemnastej przy szatni", 3)
    println(tajne)
    println(szyfruj(tajne, -3))
}
```
